````python
from anthropic import Anthropic
client = Anthropic(api_key= ANTHROPIC_API_KEY)
model = "claude-sonnet-4.5"
````

````python
def add_user_messages(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_messages(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages, system=None, stop_sequences=[]):
    params = {
        "model": "claude-sonnet-4-5",
        "max_tokens": 1000,
        "messages": messages,
        "stop_sequences": stop_sequences,
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
````

````python
messages = []

add_user_messages(messages, "Generate a very short event bridge rule as JSON.")
add_assistant_messages(messages , "```json")

text = chat(messages , stop_sequences=["```"])
````
